<p>def calculate_rsi(data, period=14):<br />
delta = data['Close'].diff()<br />
gain = (delta.where(delta &gt; 0, 0)).rolling(window=period).mean() loss
= (-delta.where(delta &lt; 0, 0)).rolling(window=period).mean() rs =
gain / loss<br />
rsi = 100 - (100 / (1 + rs))<br />
return rsi</p>
